# টোকেনাইজেশন ও এনএলপির মৌলিক ধারণা

---

ন্যাচারাল ল্যাঙ্গুয়েজ প্রসেসিং (NLP)-এর প্রথম ও অন্যতম গুরুত্বপূর্ণ ধাপ হলো টেক্সট 'নরমালাইজেশন'। কম্পিউটার যেহেতু সংখ্যা ছাড়া অন্য কোনো তথ্য সরাসরি প্রসেস করতে পারে না, তাই মানুষের ভাষাকে কম্পিউটারের বোধগম্য করতে বাক্যের শব্দগুলোকে সঠিকভাবে এনকোড করতে হয়। কোনো টেক্সট বা বাক্যকে অর্থপূর্ণ ছোট ছোট টুকরো বা শব্দে ভাগ করে সেগুলোকে সংখ্যায় রূপান্তর (Encode) করার এই প্রক্রিয়াকেই বলা হয় টোকেনাইজেশন।

### এনএলপির প্রধান দুটি মূল নীতি (Core Principles)

1. **টোকেনাইজেশন (Tokenization):** টেক্সটকে ভেঙে ছোট ছোট টুকরো বা শব্দে (Token) রূপান্তর করা।
2. **সিকোয়েন্সিং (Sequencing):** সেই শব্দগুলোকে একটি নির্দিষ্ট ধারাবাহিক সংখ্যায় সাজিয়ে কম্পিউটারের জন্য প্রসেস করতে পারে মতো করা।

### ক্যারেক্টার বনাম টোকেন এনকোডিং

ভাষার সবচেয়ে ছোট একক হলো 'অক্ষর' বা ক্যারেক্টার (যেমন: ইংরেজিতে ASCII বা বাংলায় Unicode)। কিন্তু প্রতিটি অক্ষরকে আলাদাভাবে সংখ্যায় রূপান্তর করলে শব্দের মূল অর্থ হারিয়ে যায়।

* **অক্ষর-ভিত্তিক সমস্যা:** অক্ষরকে এনকোড করলে কম্পিউটার শুধু বর্ণ চেনে, শব্দের ভাব বা অর্থ বুঝতে পারে না।
* **শব্দ-ভিত্তিক সমাধান (Tokenization):** পুরো শব্দকে একটি 'টোকেন' ধরে সংখ্যায় ম্যাপিং করলে বাক্যের অর্থ ও ব্যাকরণগত সম্পর্ক বজায় থাকে।

````python
# '\u0986' হলো 'আ' অক্ষরের ইউনিকোড মান
c='\u0986'
print(c)
````

ধরি আমরা একটা শব্দ বলছি *'কমলচরণ'*। যার অর্থ পদ্মের মতো সুন্দর পা। তিনটাকে আলাদা আলাদা করে ইউনিকোড পয়েন্ট কোড হচ্ছেঃ

````python
import numpy as np
word= 'কমলচরণ'

# প্রতিটি অক্ষরের ord মান নিয়ে array তৈরি
np.array([ord(char) for char in word])
````

সাধারণত পাইথনের বিল্ট-ইন `ord()` ফাংশন একবারে শুধু একটি অক্ষরের মান বের করে (যেমন: `ord('অ')` এর মান 2437)। পুরো NumPy অ্যারের ওপর এটি সরাসরি ব্যবহার করতে গেলে লুপের প্রয়োজন হয়।  পুরো NumPy অ্যারের ওপর একবারে কাজ করানোর জন্যই আমরা চাইলে `np.vectorize(ord)` ব্যবহার করতে পারি, যাতে আলাদা করে কোনো লুপ লিখতে না হয়।  ওয়ান-ডাইমেনশনাল  অ্যারের ক্ষেত্রে সুবিধাটি সামান্য মনে হলেও, ম্যাট্রিক্স বা মাল্টি-ডাইমেনশনাল  ডেটা প্রসেসিংয়ে এটি বেশ কাজে আসবে।

````python
import numpy as np

word = 'কমলচরণ'

# ১. শব্দটির অক্ষরগুলো নিয়ে একটি NumPy অ্যারে তৈরি
# ['ক', 'ম', 'ল', 'চ', 'র', 'ণ']
char_array = np.array(list(word))

# ২. ord ফাংশনটিকে vectorize করা
# নতুন এই v_ord ফাংশনটির ভেতরে আগে থেকেই একটি Internal Loop তৈরি করা থাকে।
v_ord = np.vectorize(ord)

# ৩. এবার সেই ফাংশনটি পুরো অ্যারের ওপর প্রয়োগ করা
ord_array = v_ord(char_array)

print(ord_array)
````

এরকম আরেকটা শব্দ হলো।  চরণকমল: চরণ বা পা-কে পদ্মের সাথে তুলনা করা।

````python
import numpy as np

word = 'চরণকমল'
char_array = np.array(list(word)) # ['চ', 'র', 'ণ','ক', 'ম', 'ল']
v_ord = np.vectorize(ord)

# ফাংশনটি পুরো অ্যারের ওপর প্রয়োগ করা
ord_array = v_ord(char_array)

print(ord_array)
````

NumPy এর মতো আলাদা করে vectorize বা for লুপ লেখার দরকার নেই। TensorFlow-এর unicode_decode স্বয়ংক্রিয়ভাবে পুরো বাংলা স্ট্রিংটিকে একবারে ডিকোড করে ইউনিকোড টেনসর (Tensor)-এ রূপান্তর করে দেয়।

````python
import tensorflow as tf

word = 'কমলচরণ'

# NumPy এর মতো আলাদা vectorize বা for-loop না লিখে সরাসরি
# পুরো টেক্সটকে একবারে ডিকোড করে ইউনিকোড টেনসরে রূপান্তর করার জন্য TensorFlow ব্যবহার করা হয়েছে।
tf_ord_tensor = tf.strings.unicode_decode(word, input_encoding='UTF-8')

print("TF Tensor   :", tf_ord_tensor)
````

Output-এ,  

*   `tf.Tensor(...)`: এটি নির্দেশ করে যে আউটপুটটি সাধারণ পাইথন লিস্ট বা NumPy অ্যারে নয়, এটি একটি TensorFlow Tensor।

*   `[2453 2478 2482 2458 2480 2467]`: এটি "কমলচরণ" শব্দটির ৬টি বাংলা অক্ষরের (ক, ম, ল, চ, র, ণ) নিজ নিজ ইউনিকোড (ASCII/Unicode Value) সংখ্যার তালিকা।

*   shape=(6,): এর অর্থ হলো এটি একটি ১D (One-dimensional) অ্যারে বা ভেক্টর, যার মধ্যে মোট ৬টি উপাদান (Element) রয়েছে।

*   dtype=int32: এটি ডেটা টাইপ নির্দেশ করে। প্রতিটি ইউনিকোড সংখ্যা এখানে 32-bit Integer (পূর্ণসংখ্যা) হিসেবে সংরক্ষিত হয়েছে।

# **ইউনিকোড এনকোডিংয়ের লিমিট ও শব্দের এনকোডিংয়ের প্রয়োজনীয়তা:**


একই অক্ষর দিয়ে গঠিত ভিন্ন ভিন্ন শব্দে অক্ষরগুলোর ইউনিকোড মান একই থাকে, কেবল তাদের অবস্থান পরিবর্তিত হয়।

*   কমলচরণঃ `[2453 2478 2482 2458 2480 2467]`
*   চরণকমলঃ `[2458 2480 2467 2453 2478 2482]`



কেবল অক্ষরের এই পজিশনাল বা ইউনিকোড মান দেখে কোনো বাক্যের অর্থ, ভাব বা কনটেক্সট  বোঝা সম্ভব নয়। কারণ ইউনিকোড কেবল অক্ষরের আইডি প্রকাশ করে, ব্যাকরণ বা শব্দের সম্পর্ক নয়।


> *মডেলকে ভাষা বা ভাবের অর্থ বোঝাতে প্রতিটি অক্ষরের বদলে পুরো শব্দকে এনকোড (Word Embedding বা Vectorization) করতে হয়। এর ফলেই মডেল বুঝতে পারে যে 'কমল' এবং 'চরণ' দুটি ভিন্ন শব্দ হওয়া সত্ত্বেও তাদের ভাব বা অর্থ একই।*



তাই সহজ ভাবে বললে বাক্যের ভাবার্থ বের করার জন্য আমাদের অক্ষর এনকোড না করে শব্দকে লেবেল করতে হবে। এই কথা বুঝার জন্য আমরা আগে পুরা একটা বাক্যে কে শব্দ দিয়ে এনকোড করে দেখি।


````python
sentences = 'আমি বই পড়তে পছন্দ করি'

# NumPy এর মতো আলাদা vectorize বা for-loop না লিখে সরাসরি
# পুরো টেক্সটকে একবারে ডিকোড করে ইউনিকোড টেনসরে রূপান্তর করার জন্য TensorFlow ব্যবহার করা হয়েছে।
tf_ord_tensor = tf.strings.unicode_decode(sentences,'UTF-8')

print("TF Tensor   :", tf_ord_tensor)
````

Output-এ, Shape দেখাচ্ছে 21 মানে বাক্যের ১৭টা অক্ষর ৪ টা স্পেস  হিসেব করে। সাথে নিজ নিজ ইউনিকোড (ASCII/Unicode Value) সংখ্যার তালিকা। আউটপুটে যেখানেই 32 দেখা যাচ্ছে, সেটি মূলত দুটি শব্দের মাঝের স্পেস (' ') নির্দেশ করছে। কারণ স্পেসের Unicode Value হলো ৩২। বাকিটা আগে একবার ব্যাখ্যা করা হয়েছে।

স্পেসের মাধ্যমে শব্দের বিভাজন বোঝা গেলেও, কেবল এই ইউনিকোড সংখ্যাগুলো দেখে বাক্যের কোনো ভাব, ব্যাকরণ বা গভীর অর্থ কিংবা Context Pattern বোঝা সম্ভব নয়।

এখানে আমরা একটা কাজ করতে পারি।  লেবেলিং!


*   'আমি' শব্দকে '1'
*   'বই' শব্দকে '2'
*   'পড়তে' শব্দকে '3'
*   'পছন্দ' হলো '4'
*   'করি' কে '5'

এভাবে লেবেলিং করে দিলাম। আর এই কাজটা করবো tokenizer দিয়ে।

````python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(num_words=10, filters='!')

sentences = ['আমি বই পড়তে পছন্দ করি!']



tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index
print(word_index)
````

*  **text প্রসেসিং মডিউল:** Keras- ভিতরে TensorFlow-এর একটি সাব-মডিউল, যা লেখা বা টেক্সট প্রসেস করার জন্য তৈরি কিছু প্রি-বিল্ট ফাংশন ও মেথডের কালেকশন।

*  **Tokenizer ক্লাস:** এটি একটি  পাইথন ক্লাস। এটি মূলত টেক্সট প্রসেসিং ঘরের এমন একটি তৈরি করা টুল, যার প্রধান কাজ হলো যেকোনো লেখা বা বাক্যকে ভেঙে ছোট ছোট টুকরো বা টোকেন কিংবা ইউনিক শব্দে ভাগ করা এবং  এবং প্রতিটি শব্দকে একটি করে নির্দিষ্ট নম্বর বা আইডি দেওয়া হয় যাতে কম্পিউটার বা এআই মডেল সহজে তা বুঝতে পারে।

এক কথায় সহজ অর্থ: `tensorflow.keras.preprocessing.text` থেকে বাক্যকে সংখ্যায় রূপান্তর করার বিশেষ মেশিনটি হল Tokenizer। যা আমরা আমাদের কোডে ব্যবহার করার জন্য import করেছি।


```

tokenizer = Tokenizer(num_words=10, filters='!')

```

**কাজ:** Tokenizer ক্লাসের একটি অবজেক্ট/ইনস্ট্যান্স তৈরি করা হচ্ছে। গুগল কোলাব বা জুপিটার নোটবুকে কোনো ভ্যারিয়েবল যেমন: tokenizer একবার মেমোরিতে তৈরি হয়ে গেলে, আমি স্ক্রিনের যে সেলই রান করি না কেন, `tokenizer.fit_on_texts()` চালু করলেই সে আগের সব হিসেব মেমোরিতে ধরে রাখে। তাই টোকেনাইজারটিকে রি-সেট করার জন্য আবার অবজেক্ট ডিক্লেয়ার করা করেছি।


```

sentences = ['আমি বই পড়তে পছন্দ করি!']

```
একটি পাইথন লিস্ট তৈরি করা হচ্ছে, যেখানে একটি বাংলা বাক্য স্ট্রিং হিসেবে রাখা হয়েছে।

*   **num_words=10:** এটি অবজেক্টের ভেতর সর্বোচ্চ ১০টি Most Frequent শব্দ ধরে রাখার লিমিট সেট করে।

*   **filters='!':** ডিফল্টভাবে টোকেনাইজারে অনেক বিরামচিহ্ন থাকে। এখানে কাস্টম ফিল্টার হিসেবে কেবল '!' দেওয়া হয়েছে। ব্যাকগ্রাউন্ডে টেক্সট প্রসেস হওয়ার সময় প্রতিটি শব্দ থেকে '!' চিহ্নটি রেগুলার এক্সপ্রেশন (Regex) দিয়ে স্ট্রিপ (Strip/Remove) করে ফেলে দেওয়া হয়।

tokenizer.fit_on_texts(sentences)

1. Lowercase & Clean: বাক্যটিকে ছোট হাতের অক্ষরে রূপান্তর করে এবং filters='!'-এর নিয়ম অনুযায়ী 'করি!' শব্দ থেকে '!' কেটে।  কেবল 'করি' শব্দটা রাখে।

2. Splitting: স্পেস ধরে বাক্যটিকে আলাদা আলাদা শব্দে বা Tokens এ রূপান্তর করে। যেমন: ['আমি', 'বই', 'পড়তে', 'পছন্দ', 'করি']

3. Frequency Counting: প্রতিটি শব্দ পুরো টেক্সটে কতবার এসেছে তার একটি ফ্রিকোয়েন্সি কাউন্টার তৈরি করে।

## 🔍 Deep Dive: ব্যাকগ্রাউন্ডে Tokenizer কীভাবে কাজ করে?

> 💡 **আমার মতো কৌতূহলী মনের জন্য (Behind the Scenes):**  
> কোনো থার্ড-পার্টি লাইব্রেরি বা এক্সট্রা মডিউল (যেমন: Keras/NLTK) ছাড়াই সম্পূর্ণ **কোর পাইথন (Core Python)** ব্যবহার করে `fit_on_texts()` মেথডটির ভেতরের মেকানিজম নিচে সহজ কোডে দেখানো হলো।

````python
# ১. কাঁচা টেক্সট (Raw Sentences)
sentences = ['আমি বই পড়তে পছন্দ করি', 'আমি বই লিখতে পছন্দ করি']

# ২. Tokenizer-এর ভেতরের লজিক (সিম্পল পাইথন কোড)
word_counts = {}

for sentence in sentences:
    words = sentence.split()  # স্পেস দিয়ে শব্দে ভাগ করা
    for word in words:
        # ব্যাকগ্রাউন্ডে শব্দের সংখ্যা গণনা করা
        word_counts[word] = word_counts.get(word, 0) + 1

# ৩. ফ্রিকোয়েন্সি অনুযায়ী ইনডেক্সিং করা
# বেশি ব্যবহৃত শব্দগুলোকে ১, ২, ৩... ইনডেক্স দেওয়া
sorted_words = sorted(word_counts, key=word_counts.get, reverse=True)
word_index = {word: index + 1 for index, word in enumerate(sorted_words)}

print("word_counts:", word_counts)
print("word_index:", word_index)
````

* ইনপুট হিসেবে প্রসেস করার জন্য মূল বাক্য দুটিকে একটি লিস্টের (List) ভেতরে নেওয়া হয়েছে।

# ২. Tokenizer-এর ভেতরের লজিক (সিম্পল পাইথন কোড)

* শব্দগুলোর সংখ্যা বা ফ্রিকোয়েন্সি জমা রাখার জন্য একটি খালি ডিকশনারি তৈরি করা হয়েছে।

`for sentence in sentences:`

* লিস্টের ভেতর থেকে প্রতিটি বাক্যকে এক এক করে লুপের মাধ্যমে প্রসেস করা হচ্ছে।

`words = sentence.split()`

* `split()` মেথড ব্যবহার করে বাক্যের স্পেস ধরে ধরে শব্দগুলোকে আলাদা করা হচ্ছে।

`for word in words:`

* আলাদা করা শব্দগুলোর ওপর আরও একটি লুপ চালিয়ে প্রতিটি শব্দকে আলাদাভাবে ট্র্যাক করা হচ্ছে।

`word_counts[word] = word_counts.get(word, 0) + 1`

* `get()` মেথড দিয়ে শব্দটি ডিকশনারিতে আগে থেকে আছে কি না তা চেক করা হচ্ছে। যদি থাকে তবে তার আগের মানের সাথে ১ যোগ হচ্ছে, আর না থাকলে মান হিসেবে ১ বসানো হচ্ছে।

# ৩. ফ্রিকোয়েন্সি অনুযায়ী ইনডেক্সিং করা

`sorted_words = sorted(word_counts, key=word_counts.get, reverse=True)`

* সবচেয়ে বেশি ব্যবহৃত শব্দগুলোকে ক্রমানুসারে (Descending Order) সাজিয়ে নেওয়া হচ্ছে।

`word_index = {word: index + 1 for index, word in enumerate(sorted_words)}`

* `enumerate()` ব্যবহার করে প্রতিটি শব্দকে ১ থেকে শুরু করে একটি  ইউনিক আইডি নম্বর দেওয়া হচ্ছে।
