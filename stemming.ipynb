# স্টেমিং (Stemming)
প্রতিটি শব্দের একটা রুট বা মূল শব্দ থাকে। এটার সাথে কিছু অক্ষর লাগিয়ে আমরা অনেক কিছু বানিয়ে ফেলি। যেমনঃ বইগুলো, বইয়ের, বইটিতে এর মূল শব্দ	বই।  বইগুলো, বইয়ের, বইটিতে এই শব্দ থেকে অতিরক্ত শব্দ বাদ দিয়ে শব্দকে বই বানিয়ে ফেলাটাকে স্টেমিং বলে। সহজ কথায়, একটি মূল শব্দ থেকে তৈরি হওয়া বিভিন্ন রূপকে একটি সাধারণ মূল শব্দে রূপান্তর করার প্রক্রিয়াই হলো স্টেমিং।

# কীভাবে কাজ করে?
স্টেমিং অ্যালগরিদম কোনো শব্দের ব্যাকরণগত অর্থ বা ভাষার অভিধান (Dictionary) চেক করে না। এটি কেবল নির্দিষ্ট কিছু নিয়ম বা অ্যালগরিদম (যেমন: Porter Stemmer, Lancaster Stemmer) অনুসরণ করে শব্দের শেষ অংশ কেটে ফেলে।

```python
from bangla_stemmer.stemmer import stemmer
word_list= ['বাড়িতে','আমাকে', 'ভালোবেসে','বইগুলো', 'বইয়ের', 'বইটিতে']
stmr = stemmer.BanglaStemmer()
stm= stmr.stem(word_list)
print(stm)
```

```python
from py_bangla_stemmer import BanglaStemmer

stemmer = BanglaStemmer()

word_list = ['বাড়িতে', 'আমাকে', 'ভালোবেসে', 'বইগুলো', 'বইয়ের', 'বইটিতে', 'জনপ্রিয়তা', 'সেটাই']

def safe_stem(w):
    try:
        return stemmer.stem(w)
    except IndexError:
        return w

stm = [safe_stem(w) for w in word_list]
print(stm)
```

```python
import nltk
from nltk.stem import PorterStemmer

# ১. অবজেক্ট তৈরি
stemmer = PorterStemmer()

word_list = ['running', 'runs', 'runner', 'eating', 'eats', 'easily', 'fairly']

# ২. প্রতিটি শব্দের ওপর স্টেমিং চালানো
stm = [stemmer.stem(word) for word in word_list]

print(stm)
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

# প্রয়োজনীয় ডাটা ডাউনলোড (প্রথমবার রান করার জন্য)
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

# ১. সাধারণ লেমাটাইজেশন (By default এটি Noun হিসেবে ধরে নেয়)
print("studies ->", lemmatizer.lemmatize("studies"))  # study

# ২. POS Tag (Parts of Speech) উল্লেখ করে দেওয়া
print("better (adj) ->", lemmatizer.lemmatize("better", pos="a")) # good
print("running (verb) ->", lemmatizer.lemmatize("running", pos="v")) # run
```
