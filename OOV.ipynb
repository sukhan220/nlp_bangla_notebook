# 'আউট অব ভোকাবুলারি' (OOV)
সহজ ভাবে বললে মডেলের শেখা Vocabulary এর বাইরের যেকোনো শব্দই হলো OOV। টোকেনাইজার ডিক্লেয়ার করার সময় oov_token প্রোপার্টি দিয়ে এটি সেট করা যায় (যেমন: Tokenizer(oov_token='<OOV>'))। আর পরবর্তীতে get_config() মেথড ব্যবহার করে দেখা যায় oov_token টি ঠিকভাবে সেট হয়েছে কি না।

````python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# [ধাপ-১] ট্রেনিং ডাটা দিয়ে টোকেনাইজার সেটআপ (oov_token যোগ করা হয়েছে)
sentences = [
    "আমি বই পড়তে পছন্দ করি।",
    "আমি বই লিখতে পছন্দ করি",
    "নতুন নতুন লেখকের মাঝে এআই এর একটা প্রভাব দেখা যাচ্ছে।"
]

tokenizer = Tokenizer(oov_token="<OOV>")  # মূল পরিবর্তনটি এখানেই!
tokenizer.fit_on_texts(sentences)
word_index = tokenizer.word_index
sentences_seq = tokenizer.texts_to_sequences(sentences)
sentences_pad= pad_sequences(sentences_seq, padding='post')





# ৩. অজানা শব্দসহ টেস্ট ডাটা
test_data = [
    'আমি আসলেই ভ্রমণ করতে পছন্দ করি বেশি',
    'আমি তেমন একটা নতুন লেখকের বই পড়ি না এখন'
]

# ৪. সিকোয়েন্স ও প্যাডিং
test_seq = tokenizer.texts_to_sequences(test_data)
padded = pad_sequences(test_seq, padding='post')

print("Word Index: ", word_index)
print("Test Sequence: ", test_seq)
print("Padded Sequence:\n", padded)
````

এখন দেখা যাচ্ছে `আউট অব ভোকাবুলারি` শব্দ গুলোর index 1।  word_index ও দেখা যাচ্ছে `'<OOV>':1`। মানে মিসিং শব্দ গুলো ইগ্নোর না করে  সিকোয়েন্স নাম্বার আসবে `1`।


``` python

```

| :-: | :--- | :--- | :--- |

| **২** | `"আমি তেমন একটা নতুন লেখকের বই পড়ি না এখন"` | `[1, ?, 12, 5, 8, 2, ?, ?, ?]` | `[1, 1, 12, 5, 8, 2, 1, 1, 1]` |

````python
print('word index', word_index)

print("Fit word Sequence: ", sentences)
print("Sequence: ", sentences_seq)
print("Test Data: ", test_data)
print(" Padded Sequence: ")
print(sentences_pad)
````
